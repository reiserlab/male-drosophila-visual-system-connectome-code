### Get neuron info

```python
neuron_df = get_special_neuron_list()
```

```python
tally = neuron_df\
    .groupby('main_groups')\
    .agg({
        'bodyId': 'count'
      , 'downstream': 'sum'
      , 'upstream': 'sum'
      , 'pre': 'sum'
      , 'post': 'sum'
      , 'instance': 'nunique'
      , 'type': 'nunique'
    })
display(tally)
```

```python
display(tally.sum(axis=0))
```
